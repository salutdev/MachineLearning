```python
from transformers import AutoTokenizer
```

```python
checkpoint = 'distilbert-base-uncased'
```

```python
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

```python
tokenized_sentences = tokenizer(raw_datasets['train'][0:3]['sentence'])
```

```python
tokenized_sentences
```

```python
#raw_datasets['train'][0:3]['sentence']
from pprint import pprint
pprint(tokenized_sentences)
```

```python
def tokenize_fn(batch):
  return tokenizer(batch['sentence'], truncation=True)
```

```python
tokenized_datasets = raw_datasets.map(tokenize_fn, batched=True)
```

```python
from transformers import TrainingArguments
```

```python
training_args = TrainingArguments('my_trainer', eval_strategy='epoch', save_strategy='epoch', num_train_epochs=1)
```

```python
from transformers import AutoModelForSequenceClassification
```

```python
model = AutoModelForSequenceClassification.from_pretrained(checkpoint, num_labels=2)
```

```python
params_before = []
for name, p in model.named_parameters():
  params_before.append(p.detach().cpu().numpy())
```

```python
from transformers import Trainer
```

```python
from datasets import load_metric
```

```python
!pip install evaluate
```

```python
import evaluate
metric = evaluate.load("glue", "sst2")
```

```python
metric.compute(predictions=[1,0,1], references=[1,0,0])
```

```python
def compute_metrics(logits_and_lanbels):
  logits, labels = logits_and_lanbels
  predictions = np.argmax(logits, axis=-1)
  return metric.compute(predictions=predictions, references=labels)
```

```python
from transformers import DataCollatorWithPadding

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

```python
trainer = Trainer(
    model,
    training_args,
    train_dataset=tokenized_datasets['train'],
    eval_dataset=tokenized_datasets['validation'],
    data_collator=data_collator,
    compute_metrics=compute_metrics)
```

```python
trainer.train()
```

```python
trainer.save_model('my_saved_model')
```
