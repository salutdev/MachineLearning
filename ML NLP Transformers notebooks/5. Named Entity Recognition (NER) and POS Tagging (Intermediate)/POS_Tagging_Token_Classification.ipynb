```python
nltk.download('universal_tagset')
```

```python
corpus = brown.tagged_sents(tagset='universal')
```

```python
corpus
```

```python

label2id = {}
list_of_sentences = []
label_id_count = 0

for sentence in corpus:
  list_of_words = []
  list_of_label_ids = []
  for word, label in sentence:
    list_of_words.append(word)
    if label not in label2id:
      label2id[label] = label_id_count
      label_id_count += 1

    list_of_label_ids.append(label2id[label])

  list_of_sentences.append({'inputs': list_of_words, 'targets': list_of_label_ids})
```

```python
id2label = {}

for k, v in label2id.items():
  id2label[v] = k
```

```python
id2label
```

```python
import json

with open('data.json', 'w') as file:
    json.dump(list_of_sentences, file)
```

```python
def align_targets(labels, word_ids):
  aligned_labels = []
  last_word = None
  for word in word_ids:
    if word is None:
      label = -100
    elif word != last_word:
      label = labels[word]
    else:
      label = -100

    aligned_labels.append(label)
    last_word = word

  return aligned_labels
```

```python
labels = data['train'][idx]['targets']
word_ids = t.word_ids()
aligned_targets = align_targets(labels, word_ids)
aligned_targets
```

```python
aligned_labels = [label_names[t] if t >= 0 else None for t in aligned_targets]
for x, y in zip(t.tokens(), aligned_labels):
  print(f"{x}\t{y}")
```

```python
def tokenize_fn(batch):
  tokenized_inputs = tokenizer(batch['inputs'], truncation=True, is_split_into_words=True)
  labels_batch = batch['targets']
  aligned_labels_batch = []
  for i, labels in enumerate(labels_batch):
    word_ids = tokenized_inputs.word_ids(i)
    aligned_labels_batch.append(align_targets(labels, word_ids))

  tokenized_inputs['labels'] = aligned_labels_batch
  return tokenized_inputs
```

```python
import numpy as np

def compute_metrics(logits_and_labels):
  logits, labels = logits_and_labels
  preds = np.argmax(logits, axis=-1)

  jagged_labels = [[t for t in label if t != -100] for label in labels]
  jagged_preds = [[p for p, t in zip(pred, targ) if t != -100] for pred, targ in zip(preds, labels)]

  flattened_labels = np.concatenate(jagged_labels).tolist()
  flattened_preds = np.concatenate(jagged_preds).tolist()

  accuracy = accuracy_metric.compute(predictions=flattened_preds, references=flattened_labels)
  # accuracy = accuracy_score(predictions=flattened_preds, references=flattened_labels)
  precision = precision_metric.compute(predictions=flattened_preds, references=flattened_labels, average='macro')
  recall = recall_metric.compute(predictions=flattened_preds, references=flattened_labels, average='macro')
  f1 = f1_metric.compute(predictions=flattened_preds, references=flattened_labels, average='macro')

  return {
      'accuracy': accuracy,
      'precision': precision,
      'recall': recall,
      'f1': f1,
  }
```

```python
from transformers import AutoModelForTokenClassification
model = AutoModelForTokenClassification.from_pretrained(checkpoint, id2label=id2label, label2id=label2id)
```

```python
from transformers import TrainingArguments
```

```python
training_args = TrainingArguments(
    'distilbert-finetuned-pos',
    # output_dir='training_dir',
    # eval_strategy='steps',
    # eval_steps=10,
    eval_strategy='epoch',
    save_strategy='epoch',
    learning_rate = 2e-5,
    num_train_epochs=1,
    weight_decay=0.01,
    # per_device_train_batch_size=16,
    # per_device_eval_batch_size=64
)
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets['train'],
    # eval_dataset=tokenized_datasets['test'].select(range(10)),
    eval_dataset=tokenized_datasets['test'],
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

trainer.train()
```

```python
trainer.save_model('my_saved_model')
```
